```python
import yfinance as yf
import matplotlib.pyplot as plt

apple = yf.Ticker("AAPL")
hist = apple.history(period="6mo")

hist["Close"].plot(title="AAPL 6month prices")
plt.xlabel("date")
plt.ylabel("close price(USD)")
plt.show()
```

```python
import yfinance as yf
import pandas as pd


# 소수점 없이 천단위 콤마로 표시
pd.options.display.float_format = '{:,.0f}'.format



def get_full_report(ticker_symbol):
    t = yf.Ticker(ticker_symbol)
    
    report = {
        "기업정보": t.info,
        "손익계산서": t.income_stmt,
        "재무상태표": t.balance_sheet,
        "현금흐름표": t.cashflow,
        "배당이력": t.dividends,
        "기관투자자": t.institutional_holders,
        "애널리스트추천": t.recommendations,
        "실적추정": t.earnings_estimate,
        "옵션만기일": t.options,
    }
    return report

report = get_full_report("AAPL")

# 각 항목 접근
print(report[""])
```
